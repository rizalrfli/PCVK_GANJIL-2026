### Tugas 1: Tampilkan citra hanya pada kombinasi kanal Merah-Biru dan kanal Hijau-Biru.

Berikut adalah kode untuk menampilkan citra dengan kombinasi kanal warna Merah-Biru dan Hijau-Biru. Kanal yang tidak digunakan (hijau untuk Merah-Biru, dan merah untuk Hijau-Biru) akan dinolkan. Kedua citra akan ditampilkan berdampingan dalam satu figur.

```python
# Pastikan imgRGB tersedia dari sel sebelumnya (jika belum, bisa dibuat dengan: imgRGB = cv.cvtColor(img, cv.COLOR_BGR2RGB))

# Citra Merah-Biru (kanal Hijau dinolkan)
img_rb = imgRGB.copy()
img_rb[:, :, 1] = 0  # Set kanal Hijau menjadi 0

# Citra Hijau-Biru (kanal Merah dinolkan)
img_gb = imgRGB.copy()
img_gb[:, :, 0] = 0  # Set kanal Merah menjadi 0

plt.figure(figsize=(15, 7))

plt.subplot(1, 2, 1)
plt.imshow(img_rb)
plt.title('Kanal Merah-Biru (Hijau dinolkan)')
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(img_gb)
plt.title('Kanal Hijau-Biru (Merah dinolkan)')
plt.axis('off')

plt.tight_layout()
plt.show()
```

### Tugas 2: Tampilkan citra dalam tiga bentuk pencerminan: vertikal, horizontal, dan keduanya.

Kode di bawah ini akan menampilkan citra asli dalam tiga bentuk pencerminan: vertikal (atas-bawah), horizontal (kiri-kanan), dan keduanya (gabungan vertikal dan horizontal). Semuanya akan disajikan dalam satu figur dengan tiga subplot yang berlabel.

```python
# Pastikan imgRGB tersedia dari sel sebelumnya

# Pencerminan vertikal (flip atas-bawah)
img_flip_vertical = cv.flip(imgRGB, 0)

# Pencerminan horizontal (flip kiri-kanan)
img_flip_horizontal = cv.flip(imgRGB, 1)

# Pencerminan keduanya (flip atas-bawah dan kiri-kanan)
img_flip_both = cv.flip(imgRGB, -1)

plt.figure(figsize=(18, 6))

plt.subplot(1, 3, 1)
plt.imshow(img_flip_vertical)
plt.title('Pencerminan Vertikal')
plt.axis('off')

plt.subplot(1, 3, 2)
plt.imshow(img_flip_horizontal)
plt.title('Pencerminan Horizontal')
plt.axis('off')

plt.subplot(1, 3, 3)
plt.imshow(img_flip_both)
plt.title('Pencerminan Keduanya')
plt.axis('off')

plt.tight_layout()
plt.show()
```

### Tugas 3: Gambarkan sebuah persegi panjang dan sebuah lingkaran pada bagian wajah/badan, lalu tambahkan teks nama Anda.

Pada tugas ini, saya akan menggunakan gambar `KTM AFRIZAL RAFLI.jpeg` yang telah diunggah. Saya akan menambahkan persegi panjang dan lingkaran pada bagian wajah/badan, serta menambahkan teks nama yang tertera di kartu identitas (`AFRIZAL RAFLI`) dengan font, ukuran, dan warna kustom.

```python
# Pastikan imgRGB tersedia dari sel sebelumnya
img_annotated = imgRGB.copy()

# Koordinat untuk rectangle (di sekitar wajah/badan)
# Asumsi wajah ada di sekitar koordinat (100, 350) dengan lebar sekitar 120 dan tinggi sekitar 150
# Sesuaikan koordinat ini jika gambar yang digunakan berbeda atau ingin menyorot bagian lain
rect_start_point = (100, 350)  # (x1, y1)
rect_end_point = (220, 500)    # (x2, y2)
rect_color = (0, 255, 255)     # Cyan (RGB)
rect_thickness = 5
cv.rectangle(img_annotated, rect_start_point, rect_end_point, rect_color, rect_thickness)

# Koordinat untuk circle (di sekitar tubuh/bahu)
# Asumsi tubuh ada di sekitar koordinat (160, 550) dengan radius 80
circle_center = (160, 550)     # (x, y)
circle_radius = 80
circle_color = (255, 0, 255)   # Magenta (RGB)
circle_thickness = 5
cv.circle(img_annotated, circle_center, circle_radius, circle_color, circle_thickness)

# Tambahkan teks nama
text = 'AFRIZAL RAFLI'
font = cv.FONT_HERSHEY_SIMPLEX
font_scale = 2
font_color = (255, 255, 0)     # Kuning (RGB)
font_thickness = 4
text_origin = (300, 300)       # (x, y) - sesuaikan posisi teks

cv.putText(img_annotated, text, text_origin, font, font_scale, font_color, font_thickness, cv.LINE_AA)

# Tampilkan gambar yang sudah dianotasi
plt.figure(figsize=(10, 8))
plt.imshow(img_annotated)
plt.title('Foto dengan Annotasi (Rectangle, Circle, dan Teks)')
plt.axis('off')
plt.show()
```
