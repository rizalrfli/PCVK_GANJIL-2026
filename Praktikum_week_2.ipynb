```python
import cv2 as cv
from google.colab.patches import cv2_imshow
from skimage import io
import matplotlib.pyplot as plt
import numpy as np

img  = cv.imread('KTM AFRIZAL RAFLI.jpeg')
cv2_imshow(img)

print("Resolusi gambar: ", img.shape [1], "x", img.shape[0])
```

```python
img = cv.imread('KTM AFRIZAL RAFLI.jpeg')
img2 = cv.cvtColor(img, cv.COLOR_BGR2RGB)
cv2_imshow(img2)
```

Menampilkan citra grayscale

```python
img_gray = cv.imread('KTM AFRIZAL RAFLI.jpeg', cv.IMREAD_GRAYSCALE)
#Tanpa cmap
plt.subplot(1,2,1)
plt.imshow(img_gray)
plt.title('Tanpa cmap')
#Dengan cmap
plt.subplot(1,2,2)
plt.imshow(img_gray, cmap='gray')
plt.title('Dengan cmap')
plt.show()
```

```python
plt.imshow(img_gray, cmap='magma')
```

Melakukan resizing

```python
imgRGB = cv.cvtColor(img,cv.COLOR_BGR2RGB)
img_output = cv.resize(imgRGB, (347,229))
cv2_imshow(img_output)
```

Melakukan flipping

```python
imgFlip=cv.flip(img_output,1)

fig = plt.figure(figsize=(10,1))
ax=fig.add_subplot(1,1,1)
ax.imshow(imgFlip)
```

Membuat bentuk geometri 2D dari opencv

```python
black_img = np.zeros(shape=(530,530,3), dtype=np.int16)
cv2_imshow(black_img)
```

```python
cv.rectangle(black_img,pt1=(290,0),pt2=(520,130),color=(0,0,255), thickness=10)
cv2_imshow(black_img)
```

```python
cv.rectangle(black_img,pt1=(250,250),pt2=(350,350),color=(255,0,255), thickness=15)
cv2_imshow(black_img)
```

```python
cv.circle(black_img,center=(150,150),radius=60,color=(0,255,0),thickness=10)
cv2_imshow(black_img)
```

```python
cv.circle(black_img,center=(400,400),radius=50,color=(0,255,0),thickness=-1)
cv.line(black_img,pt1=(0,0),pt2=(530,530),color=(255,255,255),thickness=5)
cv2_imshow(black_img)
```

```python
font = cv.FONT_HERSHEY_SIMPLEX
cv.putText(black_img,text='PCVK is fun!',org=(10,500),fontFace=font,fontScale=2,
           color=(255,255,0),thickness=3,lineType=cv.LINE_AA)
plt.imshow(black_img)
```
